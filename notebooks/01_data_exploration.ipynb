# 01 — Exploration des données

Statistiques des scènes Sentinel-1 (VV/VH dB), distribution des classes, aperçu visuel.


```python
from ml.data.dataset import FloodDataset
from ml.metrics import CLASS_NAMES

ds = FloodDataset('data/processed/synthetic', 'train', tile_size=None)
print(len(ds), 'scènes')
s = ds[0]
import numpy as np
lab = s['label'].numpy()
for i, n in enumerate(CLASS_NAMES[:3]):
    print(f'{n}: {(lab == i).mean():.1%}')
```
